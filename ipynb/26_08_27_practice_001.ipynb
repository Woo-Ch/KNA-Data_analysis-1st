- 한 엔진의 센서 값을 회차 순서대로 선 그래프로 보기

목표
- 한 엔진의 센서 값을 회차 순서대로 선 그래프로 그려 변화 방향 읽기

단계
- 2번 엔진을 골라 회차를 가로축으로 준비
- 센서 하나를 선 그래프로 그리기
- 제목·축 이름을 붙이고 값이 오르내리는 방향 읽기

예상 결과
- 회차가 늘수록 값이 서서히 오르는 곡선

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/21_cmapss_fd001_sample.csv")
```

```python
# - 2번 엔진을 골라 회차를 가로축으로 준비
eng_02 = df[df["unit_nr"] == 2]
print(eng_02.shape) # (287, 10)
print(eng_02.head())
# unit_nr  time_cycles  setting_1  setting_2  setting_3     s_2      s_3  \
# 192        2            1     0.0019    -0.0002      100.0  640.95  1585.27   
# 193        2            2    -0.0030    -0.0001      100.0  640.83  1585.76   
# 194        2            3    -0.0051     0.0002      100.0  640.17  1585.38   
# 195        2            4     0.0019    -0.0001      100.0  641.42  1585.18   
# 196        2            5    -0.0027    -0.0001      100.0  640.62  1585.38   

#          s_4     s_7    s_11  
# 192  1399.71  554.13  47.216  
# 193  1400.55  554.23  47.013  
# 194  1398.30  553.76  47.043  
# 195  1401.12  553.58  47.104  
# 196  1399.49  553.62  46.874  

# - 센서 하나를 선 그래프로 그리기
# - 제목·축 이름을 붙이고 값이 오르내리는 방향 읽기
plt.Figure(figsize=(5, 4))
plt.title("s_3")
plt.plot(eng_02["time_cycles"], eng_02["s_3"])
plt.xlabel("time_cycles")
plt.ylabel("s_3")
plt.show()
```
