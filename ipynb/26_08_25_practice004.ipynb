- 분포를 그리고 평균·표준편차·이상 설비로 해석

목표
- 분포를 그리고 평균·표준편차와 이상 설비로 해석

단계
- 온도 분포를 히스토그램으로 그리기
- 평균과 표준편차를 구해 분포의 중심·퍼짐과 비교
- 조건 필터링으로 정상 범위를 벗어난 설비 수 확인

예상 결과
- 온도 평균 70.0·표준편차 5.6, 78 이상 설비 2대

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/17_열처리.csv")
```

```python
# 온도 분포를 히스토그램으로 그리기
plt.figure(figsize=(8, 4))
plt.hist(df["온도"], bins=8)
plt.title("온도 구간별 개수 분포")
plt.xlabel("온도")
plt.ylabel("설비수")
plt.show()

# 평균과 표준편차를 구해 분포의 중심·퍼짐과 비교
print(df["온도"].mean()) # 859.5318181818183
print(df["온도"].std()) # 2.105118096148406

# 조건 필터링으로 정상 범위를 벗어난 설비 수 확인
high = df[df["온도"] >= 822]
print(len(high)) # 44
```
