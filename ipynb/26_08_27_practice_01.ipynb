목표
- 글자로 읽힌 날짜를 진짜 시간 값으로 바꾸기

단계
- 샘플을 열고 시간 열의 자료형을 먼저 확인
- 시간 변환 함수로 글자를 시간 값으로 바꾸기
- 바뀐 자료형을 다시 확인해 변환 성공을 확인

예상 결과
- 자료형이 글자에서 시간 형식으로 바뀜

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/20_engine01_timestamp_sample.csv")
```

```python
# 샘플을 열고 시간 열의 자료형을 먼저 확인
print(df.shape) # (182, 6)
print(f" 원본 타입 {df["timestamp"].dtype}") # 원본 타입 str

# 시간 변환 함수로 글자를 시간 값으로 바꾸기
df["timestamp"] = pd.to_datetime(df["timestamp"])

# 바뀐 자료형을 다시 확인해 변환 성공을 확인
print(f" 변환 후 타입 {df["timestamp"].dtype}") #  변환 후 타입 datetime64[us]
```
