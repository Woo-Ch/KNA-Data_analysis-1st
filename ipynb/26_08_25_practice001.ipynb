실습1. 막대그래프 직접 그리기
- 설비별 평균 온도를 막대로, 색·가로 막대로 변형

목표
- 범주별 값을 막대로 비교하고 색·방향을 바꿔 강조

단계
- groupby로 설비별 평균 온도를 구해 막대로 그리기
- 색 리스트로 가장 뜨거운 설비만 강조하고 너비 조절
- barh로 가로 막대로 바꾸고 제목·축 이름 붙이기

예상 결과
- 설비별 평균 온도 막대(C호기 최고), 색·가로 변형

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/17_열처리.csv")
```

```python
# groupby로 설비별 평균 온도를 구해 막대로 그리기
avg = df.groupby("배치")["온도"].mean()
print(avg) # 계산 확인
# 배치
# 로트A    858.420000
# 로트B    859.813333
# 로트C    860.421429
# Name: 온도, dtype: float64

# 색 리스트로 가장 뜨거운 설비만 강조하고 너비 조절
plt.figure(figsize=(8, 4))
plt.bar(avg.index, avg.values, color = ["gray", "gray", "red"], width=0.5)
plt.show()

# barh로 가로 막대로 바꾸고 제목·축 이름 붙이기
plt.figure(figsize=(8, 4))
plt.barh(avg.index, avg.values, color = ["gray", "gray", "red"])
plt.title("배치별 평균온도")
plt.ylabel("온도")
```
