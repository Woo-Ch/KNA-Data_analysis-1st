실습5. 산점도로 센서간 관계분석
- 온도-진동 산점도에 상태별 색·기준선 추가

목표
- 두 변수 산점도에 상태별 색과 임계 기준선 추가

단계
- 온도와 진동을 산점도로 찍어 관계 확인
- 상태를 색으로 매핑해 점검·정상을 색으로 구분
- axhline으로 진동 임계 기준선 그리기

예상 결과
- 온도↑ 진동↑ 양의 관계, 상태별 색·기준선 표시

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/17_열처리.csv")
```

```python
# 온도와 진동을 산점도로 찍어 관계 확인
plt.Figure(figsize=(8, 4))
plt.scatter(df["온도"], df["OP값"], color="red", alpha=0.5)
plt.show()

# 상태를 색으로 매핑해 점검, 정상을 색으로 구분
plt.Figure(figsize=(8, 4))
stat_colors = df["상태"].map({"정상": "green", "승온": "red"})
plt.scatter(df["온도"], df["OP값"], color=stat_colors, alpha=0.5)

# axhline으로 진동 임계 기준선 그리기
plt.axhline(y=90.0, color="red", linestyle="--") # Y기준선 작성
plt.xlabel("온도")
plt.ylabel("진동")
plt.show()
```
