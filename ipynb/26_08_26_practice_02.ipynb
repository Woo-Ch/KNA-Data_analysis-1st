실습2. bins·kde 조절
- bins로 구간 수를, kde로 분포 곡선을 조절

목표
- bins로 구간 수를, kde로 분포 곡선을 조절

단계
- bins 값을 여러 개로 바꿔 가며 분포 그리기
- 구간 수에 따라 막대가 굵어지고 잘게 나뉘는 변화 관찰
- kde 곡선을 더해 분포 모양을 부드럽게 확인

예상 결과
- bins 10·20·40에 따른 전류 분포 모양 차이

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_csv("../data/18_열처리.csv")
```

```python
sns.histplot(data=df, x="세정기", bins=10, kde=True)
plt.title("세정기 분포")
plt.xlabel("세정기")
plt.show()
```

```python
sns.histplot(data=df, x="세정기", bins=20, kde=True)
plt.title("세정기 분포")
plt.xlabel("세정기")
plt.show()
```

```python
sns.histplot(data=df, x="세정기", bins=40, kde=True)
plt.title("세정기 분포")
plt.xlabel("세정기")
plt.show()
```
