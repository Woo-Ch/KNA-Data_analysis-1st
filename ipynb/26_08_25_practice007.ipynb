```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/17_열처리_공정.csv")
```

```python
# 전체 shape 확인
print(df.shape) # (120, 8)

# 결측이 얼마나 있는지 확인
print(df.isna().sum().sum()) # 25

# 결측을 평균값으로 메꾸기
df = df.fillna(df.mean(numeric_only=True))
print(df.isna().sum().sum()) # 0
```
