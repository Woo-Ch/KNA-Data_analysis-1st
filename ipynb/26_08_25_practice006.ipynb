- 앞서 온도와 OP값(소음) 산포도를 참고해 온도와 CP값(진동) 산포도도 만들고 관계를 분석 판단해보세요.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/17_열처리.csv")
```

```python
plt.Figure(figsize=(8, 4))
stat_colors = df["상태"].map({"정상": "green", "승온": "red"})
plt.scatter(df["온도"], df["CP값"], color=stat_colors, alpha=0.5)
plt.axhline(y=0.444, color="red", linestyle="--")
plt.xlabel("온도")
plt.ylabel("진동")
plt.show()
```
