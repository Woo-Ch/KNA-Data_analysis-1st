실습2. 시간 조각 추출과 파생열
- 변환한 시간 열에서 연·월·일·시를 꺼내고 새 열 만들기

목표
- 변환한 시간 열에서 연·월·일·시 조각을 꺼내고 파생 열 만들기

단계
- 시간 접근자로 연·월·일·시를 각각 꺼내 첫 값 확인
- 시 조각을 새 열로 추가하기
- 새 열의 앞부분을 확인해 시간대별 분석을 준비

예상 결과
- 첫 행이 2024년 3월 1일 0시, 시 열 앞 세 값은 0·1·2

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/20_engine01_timestamp_sample.csv")
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])

# timestamp 컬럼이 datetime64 타입이라서 부분 뽑기 가능: year, day, hour, min ...

# 시간 접근자로 연·월·일·시를 각각 꺼내 첫 값 확인
# 방법1
print(df["timestamp"].dt.year.head(1))
# 0    2024
# Name: timestamp, dtype: int32
print(df["timestamp"].dt.month.head(1))
# 0    3
# Name: timestamp, dtype: int32
print(df["timestamp"].dt.day.head(1))
# 0    1
# Name: timestamp, dtype: int32
print(df["timestamp"].dt.hour.head(1))
# 0    0
# Name: timestamp, dtype: int32

# 방법2.
print(df["timestamp"].dt.year.iloc[1]) # 2024
print(df["timestamp"].dt.month.iloc[1]) # 3
print(df["timestamp"].dt.day.iloc[1]) # 1
print(df["timestamp"].dt.hour.iloc[1]) # 1

# 시 조각을 새 열로 추가하기
df["hour"] = df["timestamp"].dt.hour

# 새 열의 앞부분을 확인해 시간대별 분석을 준비
print(df["hour"].head(3).tolist()) # [0, 1, 2]
```
