실습4. 종합리포트—분포·박스플롯
- 한 화면에 개수·분포·박스플롯을 나란히 배치

목표
- subplots로 개수·분포·박스플롯을 한 화면에 배치

단계
- subplots로 가로 3칸을 만들기
- 각 칸에 라인별 개수·온도 분포·진동 박스플롯을 그리기
- 각 축에 제목을 달고 tight_layout으로 정리

예상 결과
- 개수·분포·박스플롯이 나란한 3칸 리포트

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_csv("../data/18_열처리.csv")
```

```python
# subplots로 가로 3칸을 만들기
fig, axs = plt.subplots(1, 3, figsize=(15, 4))

# 각 칸에 라인별 개수·온도 분포·진동 박스플롯을 그리기
sns.countplot(data=df, x="라인", hue="라인", palette="Set2", ax=axs[0], legend=False)
axs[0].set_title("라인별 개수")

sns.histplot(data=df, x="소입로온도", ax=axs[1], bins=17, kde=True)
axs[1].set_title("소입로온도 분포")

sns.boxplot(data=df, y="제어출력", ax=axs[2])
axs[2].set_title("제어출력 박스플롯")

# 각 축에 제목을 달고 tight_layout으로 정리
plt.tight_layout()

plt.show()
```
