```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_csv("../data/18_열처리.csv")
print(df.shape) # (297, 7)
df.info() # 컬럼의 성격(숫자형/범주형) 파악
print(df.isna().sum().sum()) # 0
```

```python
sns.histplot(
  data = df,
  x = "소입로온도"
)
plt.title("온도분포")
plt.xlabel("온도")
plt.show()
```

```python
# 간단한 박스플롯
sns.boxplot(data=df, y="소입로온도")
plt.title("소입로온도 박스플롯")
plt.ylabel("소입로온도")
plt.show()
```

```python
# 범주별 박스플롯
sns.boxplot(data=df, x="라인", y="건조출력")
plt.title("건조출력 박스플롯")
plt.ylabel("건조출력")
plt.xlabel("라인")
plt.show()
```

```python
# hue로 두기준 동시비교
sns.boxplot(data=df, x="라인", y="건조출력", hue="판정")
plt.title("건조출력 박스플롯")
plt.ylabel("건조출력")
plt.xlabel("라인")
plt.show()
```

```python
# countplot으로 개수 비교
sns.countplot(data=df, x="라인")
plt.title("라인별 개수")
plt.show()
```

```python
# barplot으로 평균비교
sns.barplot(data=df, x="라인", y="건조출력")
plt.show()
# 아래 그래프에서 세로선은 오차범위로 짧을수록 신뢰도 업
```

```python
# palette와 hue 함께 쓰는 법
sns.countplot(data=df, x="라인", hue="라인", palette="Set1", legend=False)
# palette= Set는 1~3까지 있고 그 외에도 다양한 종류가 있음
# 팔레트 종류는 seaborn 사이트에서 참조
# legend=False를 다는 이유는 x축에 이미 범례에 해당하는 내용들이 자리잡아서 중복 내용 표기를 금지시킨다.
plt.show()
```
