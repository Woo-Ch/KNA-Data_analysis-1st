```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 막대그래프 만들기
plt.figure(figsize=(8, 4))
plt.bar(["A호기", "B호기", "C호기"], [64, 70, 77])
plt.show()
```

```python
# 색상과 너비를 적용한 막대그래프
plt.Figure(figsize=(8, 4))
plt.bar(["A호기", "B호기", "C호기"], [164, 70, 47], color = ["red", "green", "blue"], width=0.5)
plt.show()
```

```python
# 막대그래프 가로로
plt.Figure(figsize=(8, 4))
plt.barh(["A호기", "B호기", "C호기"], [164, 70, 77], color = ["red", "green", "blue"])
plt.show()
```

```python
# 간단한 히스토그램 만들기
df = pd.read_csv("../data/17_열처리.csv")

plt.figure(figsize=(8, 4))
plt.hist(df["온도"], bins=7)
# 여기서 bins로 구간을 나눌 수 있음
plt.xlabel("온도")
plt.ylabel("설비수")
plt.show()
```

```python
# 간단한 산점도 그래프
plt.Figure(figsize=(8, 4))
plt.scatter(df["온도"], df["OP값"], color="red", alpha=0.5)
# alpha로 진한정도 설정 가능
plt.axhline(y=90.0, color="red", linestyle="--") # Y기준선 작성
plt.axvline(x=860.0, color="blue", linestyle="--") # X기준선 작성
plt.xlabel("온도")
plt.ylabel("진동")
plt.show()
```

```python
# 2x2 종합 그래프 만들기
plt.figure(figsize=(12, 12))
plt.suptitle("종합 리포트")

plt.subplot(2, 2, 1) # 가로2, 세로2인 종합그래프란의 1번영역을 사용한다는 뜻
plt.plot([1, 2, 3, 4, 5], [10, 20, 30, 40, 30])
plt.title("선그래프")

plt.subplot(2, 2, 2) # 가로2, 세로2인 종합그래프란의 2번영역을 사용한다는 뜻
plt.bar([1, 2, 3, 4, 5], [10, 20, 30, 40, 30])
plt.title("막대그래프")

plt.subplot(2, 2, 3) # 가로2, 세로2인 종합그래프란의 3번영역을 사용한다는 뜻
plt.hist([10, 20, 30, 40, 30])
plt.title("히스토그램")

plt.subplot(2, 2, 4) # 가로2, 세로2인 종합그래프란의 4번영역을 사용한다는 뜻
plt.scatter([1, 2, 3, 4, 5], [10, 20, 30, 40, 30])
plt.title("산포도")

plt.show()
```
