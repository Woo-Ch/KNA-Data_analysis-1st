실습6. 엔진 센서 시계열 그래프 그리기
- 정렬한 엔진 데이터로 회차 가로축 선 그래프 그리기

목표
- 정렬한 엔진 데이터로 회차를 가로축, 센서를 세로축으로 선 그래프 그리기

단계
- 상승 센서를 선 그래프로 그리고 제목·축 이름 붙이기
- 하강 센서도 같은 방식으로 그려 방향을 비교
- 두 그래프의 기울기 방향으로 추세를 눈으로 확인

예상 결과
- 한 센서는 우상향, 다른 센서는 우하향 곡선

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 한글깨짐 해결
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/20_cmapss_fd001_sample.csv")
```

```python
# 상승, 하강 센서
# X축(시간흐름): time_cycles
# Y축(결과)
# s_11은 47.3 -> 48.18로 상승
# s_7은 553.84 -> 546.17로 하강
eng_02 = df[df["unit_nr"] == 2].sort_values("time_cycles")

# 상승 센서를 선 그래프로 그리고 제목·축 이름 붙이기
plt.figure(figsize=(5, 3))
plt.title("2번 엔진 유닛")
plt.plot(eng_02["time_cycles"], eng_02["s_11"], label="상승센서")
plt.legend()
plt.xlabel("time_cycles")
plt.ylabel("s_11")
plt.show()

# 하강 센서도 같은 방식으로 그려 방향을 비교
plt.figure(figsize=(5, 3))
plt.title("2번 엔진 유닛")
plt.plot(eng_02["time_cycles"], eng_02["s_7"], label="하강센서")
plt.legend()
plt.xlabel("time_cycles")
plt.ylabel("s_7")
plt.show()

# 같이 그려서 비교
plt.figure(figsize=(5, 3))
plt.title("2번 엔진 유닛")
plt.plot(eng_02["time_cycles"], eng_02["s_11"], label="상승센서")
plt.plot(eng_02["time_cycles"], eng_02["s_7"], label="하강센서")
plt.legend()
plt.xlabel("time_cycles")
plt.show()
```
